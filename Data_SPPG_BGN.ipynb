```python
import requests
import pandas as pd
from google.colab import files
from IPython.display import display
import time
from io import StringIO
```

```python
# ==========================================
# SECTION 1 & 2: MASUK KE WEB & NGAMBIL DATA
# ==========================================
def scrape_sppg_bgn(max_pages=5):
    # max_pages=5 berarti kita mengambil 5 halaman pertama dulu sebagai contoh.
    # Website aslinya memiliki ratusan halaman. Kamu bisa ubah angkanya (misal max_pages=100)
    # jika ingin mengambil data lebih banyak.

    semua_data = []
    print(f"🔄 Memulai proses scraping dari {max_pages} halaman web BGN...")
    print("-" * 50)

    for page in range(1, max_pages + 1):
        # 1. MASUK KE WEB
        url = f"https://www.bgn.go.id/operasional-sppg/?page={page}"
        print(f"Membuka web: {url}")

        # Menyamar sebagai browser PC agar tidak diblokir oleh website
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/555.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/555.36'
        }

        response = requests.get(url, headers=headers)

        # 2. NGAMBIL DATA
        if response.status_code == 200:
            try:
                # pandas.read_html otomatis mendeteksi dan mengambil tabel dari halaman HTML
                tabel_web = pd.read_html(StringIO(response.text))

                if tabel_web:
                    df_page = tabel_web[0] # Mengambil tabel pertama yang ditemukan di web

                    # Menambahkan data "Link Sumber" ke dalam tabel sesuai permintaanmu
                    df_page['Link Sumber'] = url

                    # Menyimpan data halaman ini
                    semua_data.append(df_page)
                else:
                    print(f"⚠️ Tidak ditemukan tabel pada halaman {page}.")
            except ValueError:
                print(f"⚠️ Struktur tabel tidak terbaca pada halaman {page}.")
        else:
            print(f"❌ Gagal mengakses halaman {page}. Error code: {response.status_code}")

        # Jeda 1 detik antar halaman agar tidak dikira serangan DDOS/Spam oleh server BGN
        time.sleep(1)

    print("-" * 50)

    # Menggabungkan data dari semua halaman yang ditarik menjadi satu tabel utuh
    if semua_data:
        df_final = pd.concat(semua_data, ignore_index=True)
        return df_final
    else:
        return pd.DataFrame() # Mengembalikan tabel kosong jika gagal semua

# Menjalankan fungsi di atas
df_sppg = scrape_sppg_bgn(max_pages=2743)
```

```python
# ==========================================
# SECTION 3: TAMPILIN DATA
# ==========================================
if not df_sppg.empty:
    print("\n✅ Scraping selesai! Berikut adalah hasil datanya:")
    print(f"📊 Total SPPG yang berhasil ditarik: {len(df_sppg)} lokasi.")

    # Menggunakan display() bawaan IPython/Colab agar tabel terlihat rapi seperti di Excel
    display(df_sppg)
```

```python


# ==========================================
# SECTION 4: DOWNLOAD EXCEL
# ==========================================
if not df_sppg.empty:
    print("\n\u2b07\ufe0f Sedang menyiapkan file Excel...")

    # Mengonversi tabel Pandas menjadi file Excel
    nama_file = 'Data_SPPG_Nasional.xlsx'
    df_sppg.to_excel(nama_file, index=False)

    print(f"\u2705 File {nama_file} siap! Memulai proses download otomatis...")

    # Perintah Colab untuk men-trigger download ke laptop/HP-mu
    files.download(nama_file)
else:
    print("\n\u274c Proses dibatalkan karena tidak ada data yang berhasil diambil.")
```
